```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import glob

# Define the path where all CSV files are stored
#file_path = "path/to/your/csv/files/*.csv"  # Update this path
file_path = "/Users/pza/OneDrive - Oak Ridge National Laboratory/Projects/VA Project/Project/Samhsa/data_clean/*.csv"
```

```python
# Read and merge all CSV files with a 'Year' column
dataframes = []
for file in glob.glob(file_path):
    df = pd.read_csv(file)
    year = file.split("_")[-1].split(".")[0]  # Extract year from filename (format "statistics_table_2013.csv")
    df["Year"] = int(year)
    dataframes.append(df)   
```

```python
# Combine all data
data = pd.concat(dataframes)
```

# Think/plan/attempt to Suicide (ADSUITPAYR)

```python
# generating summary statistics table based on the selected predictors
df_selected = data.copy()
statistics_table = []
outcome_len = df_selected.ADSUITPAYR.value_counts().shape[0]


preds = df[['CATAG6', 'IRSEX', 'NEWRACE2', 'IRMARIT', 'EDUHIGHCAT', 'POVERTY3', 'UD5ILALANY', 'SPDPSTMON', 'SPDPSTYR', 'RCVSUTOMHT', 'AMDEYR', 'IRWRKSTAT18', 'INCOME', 'Health_Coverage','COUTYP4', 'IRPYUD5ALC','IRPYUD5MRJ', 'STMWYNORX', 'SEDWYNORX','IRPYUD5COC', 'IRPYUD5HER', 'IRPYUD5HAL', 'IRPYUD5INH', 'OXYCNANYYR', 'UD5ILLANY', 'BNGDRKMON', 'HVYDRKMON', 'IRINSUR4', 'IRIMPRESP', 'AD_MDEA6', 'AD_MDEA7' ]]

preds_decode = []
for p in preds:
    p_dict = str(p) + '_dict'
    p_decode = str(p) + '_decode'
    preds_decode.append(p_decode)

predictors = ["All", "Year"] + preds_decode + ["BMI2_decode"]
     
for predictor in predictors:
    # for the first predictor, we need to calculate the statistics for all the data
    if predictor == "All":
        df_tmp = df_selected.copy()
        subgroup = ["All", "All"]
        
        for i in range(outcome_len):
            subgroup.append(df_tmp.ADSUITPAYR.value_counts()[i])
           
        subgroup.append(df_tmp.shape[0])
        statistics_table.append(subgroup)
     
    
    # for the rest of the predictors, we need to calculate the statistics for each subgroup
    else:
        # sorted list of distinct values of the predictor
        # Ensure all values in the column are strings for consistent sorting
        distinct_values = list(np.sort(df_selected[predictor].astype(str).unique()))
        for distinct_value in distinct_values:
            subgroup = [predictor, distinct_value]

            if predictor == "Year":  # if the predictor is Year, we need to convert the distinct value to int
                df_tmp = df_selected[df_selected[predictor] == int(distinct_value)]
            else:  # if the predictor is not Year, we can use the distinct value directly
                df_tmp = df_selected[df_selected[predictor] == distinct_value]
            

            for i in range(outcome_len):
                if df_tmp.shape[0] == 0:
                    subgroup.append([0])
                else:
                    try:
                        subgroup.append(df_tmp.ADSUITPAYR.value_counts()[i])
                        
                    except KeyError:  # if certain outcome value (for example 0 or 1) does not exist in the subgroup
                        subgroup.append(0)
                        
            subgroup.append(df_tmp.shape[0])
            statistics_table.append(subgroup)
           
```

```python
#df_statistics_table = pd.DataFrame(statistics_table)
df_statistics_table = pd.DataFrame(statistics_table, columns=['Group', 'Group Value', "Y=0", "Y=1", "Total"])
```
